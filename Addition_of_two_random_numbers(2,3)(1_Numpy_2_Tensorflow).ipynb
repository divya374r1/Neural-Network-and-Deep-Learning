```python
#1.perform element-wise addition on two(2,3)tensors with random values using numpy

import numpy as np

# Create two (2,3) tensors with random values
A = np.random.rand(2, 3)
B = np.random.rand(2, 3)

# Element-wise addition
C = A + B

# Display results
print("Tensor A:\n", A)
print("\nTensor B:\n", B)
print("\nElement-wise Addition (A + B):\n", C)
```

```python
#2.perform element-wise addition on two(2,3)tensors with random values using numpy

import tensorflow as tf

# Create two (2,3) tensors with random values
A = tf.random.uniform((2, 3))
B = tf.random.uniform((2, 3))

# Element-wise addition
C = tf.add(A, B)

# Display results
print("Tensor A:\n", A)
print("\nTensor B:\n", B)
print("\nElement-wise Addition (A + B):\n", C)
```
